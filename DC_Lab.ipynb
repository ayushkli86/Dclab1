```python
import numpy as np
import matplotlib.pyplot as plt

# ---------------- UNRZ ----------------
def unrz(bits, bitrate):
    n = 200
    T = len(bits) / bitrate
    dt = T / (len(bits) * n)

    t = np.arange(0, T + dt, dt)
    x = np.zeros(len(t))

    for i in range(len(bits)):
        if bits[i] == 1:
            x[i*n:(i+1)*n] = 1
        else:
            x[i*n:(i+1)*n] = 0

    return t, x
```

```python
# ---------------- URZ ----------------
def urz(bits, bitrate):
    n = 200
    T = len(bits) / bitrate
    dt = T / (len(bits) * n)

    t = np.arange(0, T + dt, dt)
    x = np.zeros(len(t))

    for i in range(len(bits)):
        if bits[i] == 1:
            x[i*n:i*n+n//2] = 1
            x[i*n+n//2:(i+1)*n] = 0
        else:
            x[i*n:(i+1)*n] = 0

    return t, x
```

```python
# ---------------- PRZ ----------------
def prz(bits, bitrate):
    n = 200
    T = len(bits) / bitrate
    dt = T / (len(bits) * n)

    t = np.arange(0, T + dt, dt)
    x = np.zeros(len(t))

    for i in range(len(bits)):
        if bits[i] == 1:
            x[i*n:i*n+n//2] = 1
            x[i*n+n//2:(i+1)*n] = 0
        else:
            x[i*n:i*n+n//2] = -1
            x[i*n+n//2:(i+1)*n] = 0

    return t, x
```

```python

# ---------------- Manchester ----------------
def manchester(bits, bitrate):
    n = 200
    T = len(bits) / bitrate
    dt = T / (len(bits) * n)

    t = np.arange(0, T + dt, dt)
    x = np.zeros(len(t))

    for i in range(len(bits)):
        if bits[i] == 1:
            x[i*n:i*n+n//2] = 1
            x[i*n+n//2:(i+1)*n] = -1
        else:
            x[i*n:i*n+n//2] = -1
            x[i*n+n//2:(i+1)*n] = 1

    return t, x
```

```python
 #================= MAIN PROGRAM =================

bits = [1, 0, 1, 0, 0, 0, 1, 1, 0]
bitrate = 1

# UNRZ
t, x = unrz(bits, bitrate)
plt.figure(figsize=(10,3))
plt.plot(t, x, linewidth=3)
plt.grid(True)
plt.title("Unipolar NRZ")
plt.xlabel("Time")
plt.ylabel("Amplitude")
plt.ylim(-1.5, 1.5)
plt.show()

# URZ
t, x = urz(bits, bitrate)
plt.figure(figsize=(10,3))
plt.plot(t, x, linewidth=3)
plt.grid(True)
plt.title("Unipolar RZ")
plt.xlabel("Time")
plt.ylabel("Amplitude")
plt.ylim(-1.5, 1.5)
plt.show()

# PRZ
t, x = prz(bits, bitrate)
plt.figure(figsize=(10,3))
plt.plot(t, x, linewidth=3)
plt.grid(True)
plt.title("Polar RZ")
plt.xlabel("Time")
plt.ylabel("Amplitude")
plt.ylim(-1.5, 1.5)
plt.show()

# Manchester
t, x = manchester(bits, bitrate)
plt.figure(figsize=(10,3))
plt.plot(t, x, linewidth=3)
plt.grid(True)
plt.title("Manchester Encoding")
plt.xlabel("Time")
plt.ylabel("Amplitude")
plt.ylim(-1.5, 1.5)
plt.show()
```
